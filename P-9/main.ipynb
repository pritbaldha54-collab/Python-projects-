```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
class SalesDataAnalyzer:
    # CONSTRUCTOR
    def __init__(self):
        self.data = pd.DataFrame()
        self.plot = None


    # LOAD DATA
    def load_data(self):
        file_path = input("Enter the path of the dataset (CSV file) : ")

        try:
            self.data = pd.read_csv(file_path)
            print("\n✅ Dataset loaded successfully!")

        except FileNotFoundError:
            print("\n❌ File not found. Please try again.")


    # EXPLORE DATA
    def explore_data(self):

        if self.data.empty:
            print("\n❌ Data not loaded. Please load the data first.")
            return

        print("\n=== Explore Data ===")
        print("1. Display first 5 rows")
        print("2. Display last 5 rows")
        print("3. Display column names")
        print("4. Display data type")
        print("5. Display basic info")

        ch = int(input("Enter your choice : "))
        
        if ch == 1:
            print("\n",self.data.head())

        elif ch == 2:
            print("\n",self.data.tail())

        elif ch == 3:
            print("\n",list(self.data.columns))

        elif ch == 4:
            print("\n",self.data.dtypes)

        elif ch == 5:
            print("\n",self.data.info())
        else:
            print("\n❌ Invalid choice")



    # DATAFRAME OPERATIONS
    def dataframe_operations(self):
        if self.data.empty:
            print("❌ Data not loaded. Please load the data first.")
            return

        print("\n=== Perform DataFrame Operations ===")
        print("1. NumPy Array Operations")
        print("2. Mathematical Operations")
        print("3. Combine DataFrames")
        print("4. Split Data")
        print("5. Aggregating Functions")

        ch = int(input("Enter your choice : "))

        if ch == 1:
            column = input("Enter numeric column name : ")
            arr = self.data[column].to_numpy()

            print("\nArray :\n", arr)
            print("\nFirst element :\n", arr[0])
            print("\nLast element :\n", arr[-1])

        elif ch == 2:
            column = input("Enter numeric column name : ")
            arr = self.data[column].to_numpy()

            print("\nOriginal :\n", arr)
            print("\nAddition :\n", arr + 10)
            print("\nSubtraction :\n", arr - 10)
            print("\nMultiplication :\n", arr * 2)
            print("\nDivision :\n", arr / 2)

        elif ch == 3:
            df2 = self.data.head(2)
            result = pd.concat([self.data, df2])

            print("\nCombined DataFrame : \n")
            print(result)

        elif ch == 4:
            region = input("Enter region : ")
            result = self.data[self.data["Region"] == region]

            print("\nSelected Data : \n")
            print(result)

        elif ch == 5:
            column = input("Enter numeric column name : ")

            print(f"\nAggregating of {column} : ")
            print("\nSum : \n", self.data[column].sum())
            print("\nMean : \n", self.data[column].mean())
            print("\nCount : \n", self.data[column].count())
            print("\nMinimum : \n", self.data[column].min())
            print("\nMaximum : \n", self.data[column].max())

        else:
            print("\n❌ Invalid choice")


    # HANDLE MISSING DATA
    def clean_data(self):
        if self.data.empty:
            print("\n❌ Data not loaded. Please load the data first.")
            return

        print("\n=== Handle Missing Data ===")
        print("1. Display rows with missing values")
        print("2. Fill missing value with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing value with a specific value")

        ch = int(input("Enter your choice : "))

        if ch == 1:

            if self.data.isnull().values.any():
                print(self.data[self.data.isnull().any(axis=1)])
            else:
                print("\nNo missing value found.")

        elif ch == 2:
            numeric_cols = self.data.select_dtypes(include=np.number).columns

            if self.data[numeric_cols].isnull().values.any():
                self.data[numeric_cols] = self.data[numeric_cols].fillna(self.data[numeric_cols].mean())
                print("\n✅ Missing values filled with mean.")
            else:
                print("\nNo missing numeric values found.")

        elif ch == 3:
            if self.data.isnull().values.any():
                self.data.dropna(inplace=True)
                print("\n✅ Missing values rows dropped.")
            else:
                print("\nNo missing value found.")

        elif ch == 4:
            if self.data.isnull().values.any():
                val = input("Enter the value to replace missing values : ")
                self.data.fillna(val,inplace=True)
                print("\n✅ Missing values replaced.")
            else:
                print("\nNo missing value found.")

        else:
            print("\n❌ Invalid choice")



    # STATISTICAL ANALYSIS
    def stat_data(self):
        if self.data.empty:
            print("\n❌ Data not loaded. Please load the data first.")
            return

        print("\n=== Statistical analysis ===")
        print("1. Display summary statistics")
        print("2. Display correlation matrix")
        print("3. Standard deviation")
        print("4. Variance")
        print("5. Percentiles")

        ch = int(input("Enter your choice : "))

        if ch == 1:
            print("\nDescribe : \n")
            print(self.data.describe())

        elif ch == 2:
            print("\nCorrelation matrix : \n")
            print(self.data.corr(numeric_only=True))

        elif ch == 3:
            print("\nStandard deviation : \n")
            print(self.data.select_dtypes(include=np.number).std())

        elif ch == 4:
            print("\nVariance : \n")
            print(self.data.select_dtypes(include=np.number).var())

        elif ch == 5:
            num_col = self.data.select_dtypes(include=np.number)
            print("\n25% Percentiles : ",num_col.quantile(.25))
            print("50% Percentiles : ", num_col.quantile(.50))
            print("75% Percentiles : ", num_col.quantile(.75))

        else:
            print("\n❌ Invalid choice")


    # VISUALIZATION
    def visualize_data(self):
        if self.data.empty:
            print("\n❌ Data not loaded. Please load the data first.")
            return

        print("\n=== Data Visualization ===")
        print("1. Bar plot")
        print("2. Line plot")
        print("3. Scatter plot")
        print("4. Pie chart")
        print("5. Histogram")
        print("6. Stack plot")

        ch = int(input("Enter your choice : "))

        if ch == 1:
            x = input("Enter x-axis column name : ")
            y = input("Enter y-axis column name : ")
            sns.barplot(data=self.data, x=x, y=y, errorbar=None, color="teal")
            plt.tight_layout()
            self.plot = plt.gcf()
            plt.show()
            print("\n✅ Bar plot displayed successfully!")

        elif ch == 2:
            x = input("Enter x-axis column name : ")
            y = input("Enter y-axis column name : ")
            sns.lineplot(data=self.data, x=x, y=y, marker="o")
            plt.tight_layout()
            self.plot = plt.gcf()
            plt.show()
            print("\n✅ Line plot displayed successfully!")

        elif ch == 3:
            x = input("Enter x-axis column name : ")
            y = input("Enter y-axis column name : ")
            sns.scatterplot(data=self.data, x=x, y=y,)
            plt.tight_layout()
            self.plot = plt.gcf()
            plt.show()
            print("\n✅ Scatter plot displayed successfully!")

        elif ch == 4:
            product_qua = self.data.groupby("Region")["Profit"].sum()
            plt.pie(product_qua, labels=product_qua.index, autopct="%1.1f%%", startangle=90)
            plt.title("Product Quantity Distribution")
            plt.legend(loc="best")
            plt.tight_layout()
            self.plot = plt.gcf()
            plt.show()
            print("\n✅ Pie chart displayed successfully!")

        elif ch == 5:
            col = input("Enter column name for histogram : ")
            plt.hist(self.data[col], bins=10, edgecolor="black")
            plt.title("Histogram Chart")
            plt.tight_layout()
            self.plot = plt.gcf()
            plt.show()
            print("\n✅ Histogram displayed successfully!")

        elif ch == 6:
            x = input("Enter x-axis column name : ")
            y = input("Enter y-axis column name : ")
            plt.scatter(data=self.data, x=x, y=y, color='orange', edgecolors='black', s=100)
            plt.title("Stack plot Chart")
            plt.tight_layout()
            self.plot = plt.gcf()
            plt.show()
            print("\n✅ Stack plot chart displayed successfully!")

        else:
            print("\n❌ Invalid choice")

        
    # SAVE VISUALIZATION
    def save_visual(self):
        if self.data.empty:
            print("\n❌ Data not loaded. Please load the data first.")
            return

        print("\n=== Save Visualization ===")
        if self.plot is None:
            print("\n❌ No visualization available. \nPlease generate a visualization first.")
            return

        filename = input("Enter file name to save the visualization (e.g. scatter_plot.png) : ")

        self.plot.savefig(filename,dpi=300)
        print(f"\n✅ Visualization saved as {filename} successfully!")
```

```python
obj = SalesDataAnalyzer()

print("Welcome to Sales Data Analyzer!")

while True:
    print("\nMain Menu")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")

    choice = int(input("Enter your choice : "))

    if choice == 1:
        obj.load_data()

    elif choice == 2:
        obj.explore_data()

    elif choice == 3:
        obj.dataframe_operations()

    elif choice == 4:
        obj.clean_data()

    elif choice == 5:
        obj.stat_data()

    elif choice == 6:
        obj.visualize_data()

    elif choice == 7:
        obj.save_visual()

    elif choice == 8:
        print("\nThank You!")
        break

    else:
        print("n❌ Invalid choice. Please try again.")
```
