# Air Quality Dataset Analysis

**Beginner Python Data Analysis Project**

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Read Dataset

```python
df = pd.read_csv("Air Quality.csv")
df.head()
```

## 3. Check Dataset

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

## 4. Check Missing Values

```python
df.isnull().sum()
```

## 5. Check Duplicate Values

```python
df.duplicated().sum()
```

## 6. Basic Statistics

```python
df.describe()
```

## 7. Clean the Data

```python
# In this dataset, -200 means missing/invalid value.
df = df.replace(-200, np.nan)

df.isnull().sum()
```

```python
# Fill missing numeric values with the mean
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

df.isnull().sum()
```

## 8. Analyze CO(GT)

```python
df["CO(GT)"].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["CO(GT)"], bins=10,linewidth=1, edgecolor="black")

plt.xlabel("CO(GT)")
plt.ylabel("Frequency")
plt.title("Distribution of CO(GT)")
plt.show()
```

## 9. Analyze NOx(GT)

```python
df["NOx(GT)"].describe()
```

```python
plt.figure(figsize=(8,5))


plt.hist(df["NOx(GT)"], bins=10,linewidth=1, edgecolor="black")
plt.xlabel("NOx(GT)")
plt.ylabel("Frequency")
plt.title("Distribution of NOx(GT)")
plt.show()
```

## 10. Analyze NO2(GT)

```python
df["NO2(GT)"].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["NO2(GT)"], bins=10, linewidth=1, edgecolor="black")

plt.xlabel("NO2(GT)")
plt.ylabel("Frequency")
plt.title("Distribution of NO2(GT)")
plt.show()
```

## 11. Temperature Analysis

```python
df["T"].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["T"], bins=10, linewidth=1, edgecolor="black")

plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.title("Temperature Distribution")
plt.show()
```

## 12. Humidity Analysis

```python
df["RH"].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["RH"], bins=10, linewidth=1, edgecolor="black")

plt.xlabel("Relative Humidity")
plt.ylabel("Frequency")
plt.title("Humidity Distribution")
plt.show()
```

## 13. Correlation

```python
correlation = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(12,8))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
