```python
month_maxhigh = df.groupby("Month")["High"].max().astype(int)
month_maxlow = df.groupby("Month")["Low"].min().astype(int)
```

```python
plt.figure(figsize=(14, 8))

sns.lineplot(
    data=month_maxhigh, 
    label="High", 
    linewidth=2.5, 
    color="#2ca02c",     
    marker="o",          
    markersize=7
)
sns.lineplot(
    data=month_maxlow, 
    label="Low", 
    linewidth=2.5, 
    color="#d62728",     
    marker="o", 
    markersize=7
)

plt.title("ICICI High vs Low Price - 2020", fontsize=20, fontweight="bold")
plt.xlabel("Month", fontsize=13, fontweight="bold")
plt.ylabel("Price", fontsize=13, fontweight="bold")

plt.legend()
plt.grid(linestyle="--", linewidth=0.7, alpha=0.7)

plt.show()
```

## 📊 Additional Data Analysis

A few extra charts to make the analysis more complete and presentation-ready.

```python
# Monthly trading volume
monthly_volume = df.groupby(df['Date'].dt.to_period('M'))['Volume'].mean().reset_index()
monthly_volume['Date'] = monthly_volume['Date'].astype(str)

plt.figure(figsize=(12, 5))
sns.barplot(data=monthly_volume, x='Date', y='Volume', color='steelblue')
plt.title('Average Monthly Trading Volume', fontsize=16, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Average Volume')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Key insight:** This chart shows the months with relatively higher and lower trading activity.

```python
# Daily price range: High - Low
df['Price_Range'] = df['High'] - df['Low']

plt.figure(figsize=(12, 5))
sns.lineplot(data=df, x='Date', y='Price_Range', linewidth=1.8)
plt.title('Daily Price Range (High - Low)', fontsize=16, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Price Range')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Key insight:** A larger price range indicates stronger movement between the day's highest and lowest prices.

```python
# Distribution of closing prices
plt.figure(figsize=(12, 5))
sns.histplot(df['Close'].dropna(), bins=30, kde=True)
plt.title('Distribution of Closing Prices', fontsize=16, fontweight='bold')
plt.xlabel('Closing Price')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

**Key insight:** The distribution shows where closing prices occur most frequently in the dataset.

## 🧠 Short Key Data Analysis

- **Price:** The closing-price trend shows the overall movement of ICICI Bank stock during the selected period.
- **Volume:** Trading volume helps identify periods of higher or lower market activity.
- **Price range:** The High-Low range highlights days with larger price movements.
- **Distribution:** The closing-price distribution shows the most common price levels in the dataset.
- **Overall:** The charts together provide a simple view of price, activity, and volatility patterns.
