# World Happiness Report Analysis

**Beginner Python Data Analysis Project**

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Read Dataset

```python
df = pd.read_csv("World Happiness Report.csv")
df.head()
```

## 3. Check Dataset

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

## 4. Check Missing Values

```python
df.isnull().sum()
```

## 5. Check Duplicate Values

```python
df.duplicated().sum()
```

## 6. Basic Statistics

```python
df.describe()
```

## 7. Happiness Score Analysis

```python
df["Happiness Score"].describe()
```

```python
plt.hist(df["Happiness Score"].dropna(), bins=10, linewidth=1, edgecolor="black")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.title("Happiness Score Distribution")
plt.show()
```

## 8. Happiness Rank

```python
df["Happiness Rank"].describe()
```

## 9. Top 10 Happiest Countries

```python
top_10 = df.sort_values("Happiness Score", ascending=False).head(10)
top_10[[ "Country", "Happiness Score" ]]
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=top_10["Country"], y=top_10["Happiness Score"], palette="viridis")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.title("Top 10 Happiest Countries")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 10. Relationship Between Happiness and Another Variable

```python
plt.scatter(df["Economy (GDP per Capita)"], df["Happiness Score"])
plt.xlabel("Economy (GDP per Capita)")
plt.ylabel("Happiness Score")
plt.title("Economy (GDP per Capita) vs Happiness Score")
plt.show()
```

## 11. Correlation

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10, 7))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
