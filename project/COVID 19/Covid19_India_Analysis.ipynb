## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Read Dataset

```python
df = pd.read_csv("Covid-19 India.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

## 6. Basic Statistics

```python
df.describe()
```

## 7. Total Cases by State

```python
state_cases = df.sort_values("Total Cases", ascending=False)

state_cases[["State/UTs", "Total Cases"]].head(10)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=state_cases["State/UTs"].head(10), y=state_cases["Total Cases"].head(10), palette="muted")
plt.xlabel("State / UT")
plt.ylabel("Total Cases")
plt.title("Top 10 States by Total Cases")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 8. Active Cases by State

```python
active_cases = df.sort_values("Active", ascending=False)

active_cases[["State/UTs", "Active"]].head(10)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=active_cases["State/UTs"].head(10), y=active_cases["Active"].head(10), palette="viridis")
plt.xlabel("State / UT")
plt.ylabel("Active Cases")
plt.title("Top 10 States by Active Cases")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 9. Deaths by State

```python
death_cases = df.sort_values("Deaths", ascending=False)

death_cases[["State/UTs", "Deaths"]].head(10)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=death_cases["State/UTs"].head(10), y=death_cases["Deaths"].head(10), palette="rocket")
plt.xlabel("State / UT")
plt.ylabel("Deaths")
plt.title("Top 10 States by Deaths")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 10. Discharged Cases by State

```python
discharged_cases = df.sort_values("Discharged", ascending=False)

discharged_cases[["State/UTs", "Discharged"]].head(10)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=discharged_cases["State/UTs"].head(10), y=discharged_cases["Discharged"].head(10), palette="mako")
plt.xlabel("State / UT")
plt.ylabel("Discharged Cases")
plt.title("Top 10 States by Discharged Cases")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 11. Active, Discharged and Death Cases

```python
top_states = df.sort_values("Total Cases", ascending=False).head(10)

top_states.set_index("State/UTs")[["Active", "Discharged", "Deaths"]].plot(
    kind="bar", figsize=(12,6)
)
plt.xlabel("State / UT")
plt.ylabel("Number of Cases")
plt.title("Active, Discharged and Death Cases")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 12. Active Ratio by State

```python
ratio = df.sort_values("Active Ratio", ascending=False).head(10)

ratio[["State/UTs", "Active Ratio"]]
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=ratio["State/UTs"], y=ratio["Active Ratio"], palette="Spectral")
plt.xlabel("State / UT")
plt.ylabel("Active Ratio")
plt.title("Top 10 States by Active Ratio")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 13. Discharge Ratio by State

```python
ratio = df.sort_values("Discharge Ratio", ascending=False).head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=ratio["State/UTs"], y=ratio["Discharge Ratio"], palette="RdBu")
plt.xlabel("State / UT")
plt.ylabel("Discharge Ratio")
plt.title("Top 10 States by Discharge Ratio")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 14. Death Ratio by State

```python
ratio = df.sort_values("Death Ratio", ascending=False).head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=ratio["State/UTs"], y=ratio["Death Ratio"], palette="Greens")
plt.xlabel("State / UT")
plt.ylabel("Death Ratio")
plt.title("Top 10 States by Death Ratio")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 15. Population by State

```python
population = df.sort_values("Population", ascending=False).head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=population["State/UTs"], y=population["Population"], palette="mako")
plt.xlabel("State / UT")
plt.ylabel("Population")
plt.title("Top 10 States by Population")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## 16. Total Cases vs Population

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Population"], df["Total Cases"])
plt.xlabel("Population")
plt.ylabel("Total Cases")
plt.title("Population vs Total Cases")
plt.show()
```

## 17. Total Cases vs Deaths

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Total Cases"], df["Deaths"])
plt.xlabel("Total Cases")
plt.ylabel("Deaths")
plt.title("Total Cases vs Deaths")
plt.show()
```

## 18. Total Cases Distribution

```python
plt.hist(df["Total Cases"], bins=10,  linewidth=1, edgecolor="black")
plt.xlabel("Total Cases")
plt.ylabel("Number of States / UTs")
plt.title("Distribution of Total Cases")
plt.show()
```

## 19. Correlation

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10,7))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
